```python
# ============================================================
# UNITS: N, mm, MPa
# ============================================================

# ============================================================
# GEOMETRY
# ============================================================

b = 6300.0
h = 600.0

c = 100.0

y_bottom = 50.0
y_tie    = 250.0
y_top    = 550.0


# ============================================================
# MATERIALS
# ============================================================

# Concrete
matC = TS.Material(
    30000,     # Ec
    30,        # fck
    0.003,     # epsilon_c
    -1         # compression
)

# Steel
matT = TS.Material(
    205000,    # Es
    435,       # fsy
    0.08,      # epsilon_s
    1          # tension
)


# ============================================================
# OUTER GEOMETRY
# ============================================================

PG1 = TS.Point(x=0,    y=0, z=0)
PG2 = TS.Point(x=b,    y=0, z=0)
PG3 = TS.Point(x=b,    y=h, z=0)
PG4 = TS.Point(x=0,    y=h, z=0)


# ============================================================
# OPENING
# ============================================================

PG5 = TS.Point(x=450,  y=350, z=0)
PG6 = TS.Point(x=1650, y=350, z=0)
PG7 = TS.Point(x=1650, y=500, z=0)
PG8 = TS.Point(x=450,  y=500, z=0)


# ============================================================
# OUTER POLYGON
# ============================================================

Polygon = TS.Polygon()

Polygon.addPoint(PG1)
Polygon.addPoint(PG4)
Polygon.addPoint(PG3)
Polygon.addPoint(PG2)


# ============================================================
# OPENING POLYGON
# ============================================================

Opening = TS.Polygon()

Opening.addPoint(PG5)
Opening.addPoint(PG8)
Opening.addPoint(PG7)
Opening.addPoint(PG6)


# ============================================================
# DEFAULT DOF
# ============================================================

DOF0 = TS.DOF(0, 0, 0, 0, 0, 0)


# ============================================================
# STRUCTURAL NODES
# ============================================================
#
# Numbering:
#
# bottom       -> left to right
# intermediate -> left to right
# top          -> left to right
#
# ============================================================


# ============================================================
# BOTTOM NODES
# y = 50 mm
# ============================================================

P1  = TS.Point(x=150,  y=y_bottom, z=0)   # left support
P2  = TS.Point(x=400,  y=y_bottom, z=0)
P3  = TS.Point(x=600,  y=y_bottom, z=0)
P4  = TS.Point(x=800,  y=y_bottom, z=0)
P5  = TS.Point(x=1000, y=y_bottom, z=0)
P6  = TS.Point(x=1200, y=y_bottom, z=0)
P7  = TS.Point(x=1400, y=y_bottom, z=0)
P8  = TS.Point(x=1600, y=y_bottom, z=0)
P9  = TS.Point(x=1800, y=y_bottom, z=0)

P10 = TS.Point(x=2250, y=y_bottom, z=0)
P11 = TS.Point(x=2700, y=y_bottom, z=0)

# NOTE:
# There was a point at x = 3150 in your original code,
# but its Node was commented out.
#
# P12 = TS.Point(x=3150, y=y_bottom, z=0)

P12 = TS.Point(x=3600, y=y_bottom, z=0)
P13 = TS.Point(x=4050, y=y_bottom, z=0)
P14 = TS.Point(x=4500, y=y_bottom, z=0)
P15 = TS.Point(x=4950, y=y_bottom, z=0)
P16 = TS.Point(x=5400, y=y_bottom, z=0)

P17 = TS.Point(x=6150, y=y_bottom, z=0)   # right support


# ============================================================
# INTERMEDIATE NODES
# y = 250 mm
# ============================================================

P18 = TS.Point(x=250,  y=y_tie, z=0)
P19 = TS.Point(x=400,  y=y_tie, z=0)
P20 = TS.Point(x=600,  y=y_tie, z=0)
P21 = TS.Point(x=800,  y=y_tie, z=0)
P22 = TS.Point(x=1000, y=y_tie, z=0)
P23 = TS.Point(x=1200, y=y_tie, z=0)
P24 = TS.Point(x=1400, y=y_tie, z=0)
P25 = TS.Point(x=1600, y=y_tie, z=0)
P26 = TS.Point(x=1800, y=y_tie, z=0)


# ============================================================
# TOP NODES
# y = 550 mm
# ============================================================

P27 = TS.Point(x=400,  y=y_top, z=0)
P28 = TS.Point(x=2250, y=y_top, z=0)
P29 = TS.Point(x=2700, y=y_top, z=0)
P30 = TS.Point(x=3150, y=y_top, z=0)
P31 = TS.Point(x=3600, y=y_top, z=0)
P32 = TS.Point(x=4050, y=y_top, z=0)
P33 = TS.Point(x=4500, y=y_top, z=0)
P34 = TS.Point(x=4950, y=y_top, z=0)
P35 = TS.Point(x=5400, y=y_top, z=0)


# ============================================================
# CREATE NODES
# ============================================================

N1  = TS.Node(P1,  DOF0)
N2  = TS.Node(P2,  DOF0)
N3  = TS.Node(P3,  DOF0)
N4  = TS.Node(P4,  DOF0)
N5  = TS.Node(P5,  DOF0)
N6  = TS.Node(P6,  DOF0)
N7  = TS.Node(P7,  DOF0)
N8  = TS.Node(P8,  DOF0)
N9  = TS.Node(P9,  DOF0)

N10 = TS.Node(P10, DOF0)
N11 = TS.Node(P11, DOF0)
N12 = TS.Node(P12, DOF0)
N13 = TS.Node(P13, DOF0)
N14 = TS.Node(P14, DOF0)
N15 = TS.Node(P15, DOF0)
N16 = TS.Node(P16, DOF0)
N17 = TS.Node(P17, DOF0)

N18 = TS.Node(P18, DOF0)
N19 = TS.Node(P19, DOF0)
N20 = TS.Node(P20, DOF0)
N21 = TS.Node(P21, DOF0)
N22 = TS.Node(P22, DOF0)
N23 = TS.Node(P23, DOF0)
N24 = TS.Node(P24, DOF0)
N25 = TS.Node(P25, DOF0)
N26 = TS.Node(P26, DOF0)

N27 = TS.Node(P27, DOF0)
N28 = TS.Node(P28, DOF0)
N29 = TS.Node(P29, DOF0)
N30 = TS.Node(P30, DOF0)
N31 = TS.Node(P31, DOF0)
N32 = TS.Node(P32, DOF0)
N33 = TS.Node(P33, DOF0)
N34 = TS.Node(P34, DOF0)
N35 = TS.Node(P35, DOF0)


# ============================================================
# SUPPORTS
# ============================================================

# Left support
S1 = TS.Support(
    N1,
    TS.DOF(1, 1, 1, 0, 0, 0)
)

# Right support
S2 = TS.Support(
    N17,
    TS.DOF(0, 1, 0, 0, 0, 0)
)
```

```python
T2 = TS.TrussSystem()

# ============================================================
# SEARCH SPACE
# ============================================================

T2.addPolygontoSS(Polygon)

T2.addNode([
    # Bottom - left to right
    N1, N2, N3, N4, N5, N6, N7, N8, N9,
    N10, N11, N12, N13, N14, N15, N16, N17,

    # Intermediate - left to right
    N18, N19, N20, N21, N22,
    N23, N24, N25, N26,

    # Top - left to right
    N27, N28, N29, N30, N31,
    N32, N33, N34, N35
])

T2.addMaterial(matC)
T2.addMaterial(matT)


# ============================================================
# EDGES
# ============================================================


# ============================================================
# 1. BOTTOM TIE
# ============================================================
#
# N1 ---- N2 ---- ... ---- N17
#
# ============================================================

e1  = TS.Edge(N1,  N2,  matT)
e2  = TS.Edge(N2,  N3,  matT)
e3  = TS.Edge(N3,  N4,  matT)
e4  = TS.Edge(N4,  N5,  matT)
e5  = TS.Edge(N5,  N6,  matT)
e6  = TS.Edge(N6,  N7,  matT)
e7  = TS.Edge(N7,  N8,  matT)
e8  = TS.Edge(N8,  N9,  matT)

e9  = TS.Edge(N9,  N10, matT)

e10 = TS.Edge(N10, N11, matT)
e11 = TS.Edge(N11, N12, matT)
e12 = TS.Edge(N12, N13, matT)
e13 = TS.Edge(N13, N14, matT)
e14 = TS.Edge(N14, N15, matT)
e15 = TS.Edge(N15, N16, matT)
e16 = TS.Edge(N16, N17, matT)


# ============================================================
# 2. TOP COMPRESSION CHORD
# ============================================================

e17 = TS.Edge(N27, N28, matC)

e18 = TS.Edge(N28, N29, matC)
e19 = TS.Edge(N29, N30, matC)
e20 = TS.Edge(N30, N31, matC)
e21 = TS.Edge(N31, N32, matC)
e22 = TS.Edge(N32, N33, matC)
e23 = TS.Edge(N33, N34, matC)
e24 = TS.Edge(N34, N35, matC)


# ============================================================
# 3. INTERMEDIATE HORIZONTAL TIE
# ============================================================

e25 = TS.Edge(N18, N19, matT)
e26 = TS.Edge(N19, N20, matT)
e27 = TS.Edge(N20, N21, matT)
e28 = TS.Edge(N21, N22, matT)
e29 = TS.Edge(N22, N23, matT)
e30 = TS.Edge(N23, N24, matT)
e31 = TS.Edge(N24, N25, matT)
e32 = TS.Edge(N25, N26, matT)


# ============================================================
# 4. LEFT END
# ============================================================

# Left support -> intermediate
e33 = TS.Edge(N1, N18, matC)

# Intermediate -> top
e34 = TS.Edge(N18, N27, matC)

# Intermediate -> bottom
# old e901
e35 = TS.Edge(N18, N2, matC)


# ============================================================
# 5. VERTICAL AT x = 400
# ============================================================

e36 = TS.Edge(N27, N19, matT)
e37 = TS.Edge(N19, N2,  matT)


# ============================================================
# 6. VERTICAL TIES - DENSE REGION
# ============================================================

# x = 600
e38 = TS.Edge(N20, N3, matT)

# x = 800
e39 = TS.Edge(N21, N4, matT)

# x = 1000
e40 = TS.Edge(N22, N5, matT)

# x = 1200
e41 = TS.Edge(N23, N6, matT)

# x = 1400
e42 = TS.Edge(N24, N7, matT)

# x = 1600
e43 = TS.Edge(N25, N8, matT)

# x = 1800
e44 = TS.Edge(N26, N9, matT)


# ============================================================
# 7. DIAGONAL STRUTS - DENSE REGION
# ============================================================

e45 = TS.Edge(N2, N20, matC)
e46 = TS.Edge(N3, N21, matC)
e47 = TS.Edge(N4, N22, matC)
e48 = TS.Edge(N5, N23, matC)

e49 = TS.Edge(N6, N24, matC)
e50 = TS.Edge(N7, N25, matC)
e51 = TS.Edge(N8, N26, matC)

# Last diagonal toward upper chord
e52 = TS.Edge(N9, N28, matC)


# ============================================================
# 8. TRANSITION AFTER OPENING
# ============================================================

# x = 2250
e53 = TS.Edge(N10, N28, matT)


# ============================================================
# 9. RIGHT REGION - DIAGONAL STRUTS
# ============================================================

# 2250 bottom -> 2700 top
e54 = TS.Edge(N10, N29, matC)

# 2700 bottom -> 3150 top
e55 = TS.Edge(N11, N30, matC)

# 3150 top -> 3600 bottom
e56 = TS.Edge(N30, N12, matC)

# 3600 top -> 4050 bottom
e57 = TS.Edge(N31, N13, matC)

# 4050 top -> 4500 bottom
e58 = TS.Edge(N32, N14, matC)

# 4500 top -> 4950 bottom
e59 = TS.Edge(N33, N15, matC)

# 4950 top -> 5400 bottom
e60 = TS.Edge(N34, N16, matC)

# 5400 top -> right support
e61 = TS.Edge(N35, N17, matC)


# ============================================================
# 10. RIGHT VERTICAL TIES
# ============================================================

# x = 2700
# THIS BAR WAS CREATED AS e71 BUT WAS NOT ADDED IN THE OLD CODE
e62 = TS.Edge(N29, N11, matT)

# x = 3600
e63 = TS.Edge(N31, N12, matT)

# x = 4050
e64 = TS.Edge(N32, N13, matT)

# x = 4500
e65 = TS.Edge(N33, N14, matT)

# x = 4950
e66 = TS.Edge(N34, N15, matT)

# x = 5400
e67 = TS.Edge(N35, N16, matT)


# ============================================================
# ADD EDGES
# ============================================================

T2.addEdge([
    # Bottom tie
    e1, e2, e3, e4, e5, e6, e7, e8,
    e9, e10, e11, e12, e13, e14, e15, e16,

    # Top chord
    e17, e18, e19, e20,
    e21, e22, e23, e24,

    # Intermediate tie
    e25, e26, e27, e28,
    e29, e30, e31, e32,

    # Left region
    e33, e34, e35,

    # Vertical x = 400
    e36, e37,

    # Dense vertical ties
    e38, e39, e40, e41,
    e42, e43, e44,

    # Dense diagonal struts
    e45, e46, e47, e48,
    e49, e50, e51, e52,

    # Transition
    e53,

    # Right diagonal struts
    e54, e55, e56, e57,
    e58, e59, e60, e61,

    # Right vertical ties
    e62, e63, e64, e65, e66, e67
])

# ============================================================
# EXTERNAL FORCES
# ============================================================

q = 30.0  # N/mm

DOF_F = TS.DOF(0, 1, 0, 0, 0, 0)


# ============================================================
# LOADS ON TOP NODES
# ============================================================
#
# N27 = x = 400
# N28 = x = 2250
# N29 = x = 2700
# N30 = x = 3150
# N31 = x = 3600
# N32 = x = 4050
# N33 = x = 4500
# N34 = x = 4950
# N35 = x = 5400
#
# ============================================================

Force1 = TS.Force_ext(
    N27,
    [0, -35250, 0, 0, 0, 0],
    DOF_F
)

Force2 = TS.Force_ext(
    N28,
    [0, -34500, 0, 0, 0, 0],
    DOF_F
)

Force3 = TS.Force_ext(
    N29,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force4 = TS.Force_ext(
    N30,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force5 = TS.Force_ext(
    N31,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force6 = TS.Force_ext(
    N32,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force7 = TS.Force_ext(
    N33,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force8 = TS.Force_ext(
    N34,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force9 = TS.Force_ext(
    N35,
    [0, -29250, 0, 0, 0, 0],
    DOF_F
)


# ============================================================
# ADD FORCES TO T2
# ============================================================

T2.addForce(Force1)
T2.addForce(Force2)
T2.addForce(Force3)
T2.addForce(Force4)
T2.addForce(Force5)
T2.addForce(Force6)
T2.addForce(Force7)
T2.addForce(Force8)
T2.addForce(Force9)

# add Supports
T2.addSupport(S1)
T2.addSupport(S2)

T2.solveTruss(True)
print('Is strut-and-tie model in equilibrium? ', T2.checkEquilibrium())
```

```python
print("\n==============================")
print("ACTUAL LOADS IN T2")
print("==============================")

total_Fy = 0
total_M = 0

for i, F in enumerate(T2.ForceList):

    x = F.node.point.x
    y = F.node.point.y

    Fx = F.Force_magnitude[0]
    Fy = F.Force_magnitude[1]

    total_Fy += Fy
    total_M += Fy * x

    print(
        f"Force {i+1:2d}: "
        f"x = {x:7.1f} mm | "
        f"y = {y:6.1f} mm | "
        f"Fy = {Fy/1000:8.3f} kN"
    )

print("\nTOTAL Fy =", total_Fy/1000, "kN")

x_resultant = total_M / total_Fy

print("Resultant position x =", x_resultant, "mm")
```
