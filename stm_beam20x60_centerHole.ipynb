# Example of Helmi Ghourabi

```python
# ============================================================
# UNITS: N, mm, MPa
# ============================================================


# ============================================================
# GEOMETRY
# ============================================================

b = 6300.0  # mm
h = 600.0   # mm

c = 50.0    # mm

y_load = h - c          # 550 mm
y_hole = h - c - 150.0  # 400 mm


# ============================================================
# OUTER GEOMETRY
# ============================================================

PG1 = TS.Point(x=0, y=0, z=0)
PG2 = TS.Point(x=b, y=0, z=0)
PG3 = TS.Point(x=b, y=h, z=0)
PG4 = TS.Point(x=0, y=h, z=0)


# ============================================================
# OPENING GEOMETRY
#
# x = 2550 -> 3750 mm
# y = 150  -> 270 mm
# ============================================================

PG5 = TS.Point(x=2550, y=150, z=0)
PG6 = TS.Point(x=3750, y=150, z=0)
PG7 = TS.Point(x=3750, y=270, z=0)
PG8 = TS.Point(x=2550, y=270, z=0)


# ============================================================
# MATERIALS
# ============================================================

# Concrete
matC = TS.Material(
    30000,
    20,
    0.003,
    -1
)

# Steel
matT = TS.Material(
    205000,
    435,
    0.08,
    1
)


# ============================================================
# DEFAULT DOF
# ============================================================

DOF0 = TS.DOF(0, 0, 0, 0, 0, 0)


# ============================================================
# OUTER POLYGON
# ============================================================

Polygon = TS.Polygon()

Polygon.addPoint(PG1)
Polygon.addPoint(PG4)
Polygon.addPoint(PG3)
Polygon.addPoint(PG2)


# ============================================================
# OPENING POLYGON
# ============================================================

Opening = TS.Polygon()

Opening.addPoint(PG5)
Opening.addPoint(PG8)
Opening.addPoint(PG7)
Opening.addPoint(PG6)


# ============================================================
# STRUCTURAL POINTS
# ============================================================
#
# Numbering:
#
# P1  -> P2  : supports
# P3  -> P19 : top nodes
# P20 -> P35 : bottom / internal tie nodes
#
# ============================================================


# ============================================================
# 1. SUPPORT POINTS
# y = 50 mm
# ============================================================

P1 = TS.Point(
    x=150,
    y=c,
    z=0
)

P2 = TS.Point(
    x=6150,
    y=c,
    z=0
)


# ============================================================
# 2. TOP POINTS - LOAD INTRODUCTION
# y = 550 mm
# ============================================================

# Left region
P3 = TS.Point(x=675,  y=y_load, z=0)
P4 = TS.Point(x=1125, y=y_load, z=0)
P5 = TS.Point(x=1575, y=y_load, z=0)
P6 = TS.Point(x=2025, y=y_load, z=0)

# Central region
P7  = TS.Point(x=2550, y=y_load, z=0)
P8  = TS.Point(x=2700, y=y_load, z=0)
P9  = TS.Point(x=2850, y=y_load, z=0)
P10 = TS.Point(x=3000, y=y_load, z=0)
P11 = TS.Point(x=3150, y=y_load, z=0)
P12 = TS.Point(x=3300, y=y_load, z=0)
P13 = TS.Point(x=3450, y=y_load, z=0)
P14 = TS.Point(x=3600, y=y_load, z=0)
P15 = TS.Point(x=3750, y=y_load, z=0)

# Right region
P16 = TS.Point(x=4275, y=y_load, z=0)
P17 = TS.Point(x=4725, y=y_load, z=0)
P18 = TS.Point(x=5175, y=y_load, z=0)
P19 = TS.Point(x=5625, y=y_load, z=0)


# ============================================================
# 3. BOTTOM / INTERNAL TIE POINTS
# ============================================================


# ============================================================
# LEFT REGION
# y = 50 mm
# ============================================================

P20 = TS.Point(x=675,  y=c, z=0)
P21 = TS.Point(x=1125, y=c, z=0)
P22 = TS.Point(x=1575, y=c, z=0)
P23 = TS.Point(x=2025, y=c, z=0)


# ============================================================
# CENTRAL REGION
# y = 400 mm
# ============================================================

P24 = TS.Point(x=2550, y=y_hole, z=0)
P25 = TS.Point(x=2700, y=y_hole, z=0)
P26 = TS.Point(x=2850, y=y_hole, z=0)
P27 = TS.Point(x=3000, y=y_hole, z=0)

# No node at x = 3150 mm

P28 = TS.Point(x=3300, y=y_hole, z=0)
P29 = TS.Point(x=3450, y=y_hole, z=0)
P30 = TS.Point(x=3600, y=y_hole, z=0)
P31 = TS.Point(x=3750, y=y_hole, z=0)


# ============================================================
# RIGHT REGION
# y = 50 mm
# ============================================================

P32 = TS.Point(x=4275, y=c, z=0)
P33 = TS.Point(x=4725, y=c, z=0)
P34 = TS.Point(x=5175, y=c, z=0)
P35 = TS.Point(x=5625, y=c, z=0)


# ============================================================
# CREATE STRUCTURAL NODES
# ============================================================


# ============================================================
# SUPPORT NODES
# ============================================================

N1 = TS.Node(P1, DOF0)   # left support
N2 = TS.Node(P2, DOF0)   # right support


# ============================================================
# TOP NODES
# ============================================================

N3  = TS.Node(P3,  DOF0)
N4  = TS.Node(P4,  DOF0)
N5  = TS.Node(P5,  DOF0)
N6  = TS.Node(P6,  DOF0)

N7  = TS.Node(P7,  DOF0)
N8  = TS.Node(P8,  DOF0)
N9  = TS.Node(P9,  DOF0)
N10 = TS.Node(P10, DOF0)
N11 = TS.Node(P11, DOF0)
N12 = TS.Node(P12, DOF0)
N13 = TS.Node(P13, DOF0)
N14 = TS.Node(P14, DOF0)
N15 = TS.Node(P15, DOF0)

N16 = TS.Node(P16, DOF0)
N17 = TS.Node(P17, DOF0)
N18 = TS.Node(P18, DOF0)
N19 = TS.Node(P19, DOF0)


# ============================================================
# BOTTOM / INTERNAL TIE NODES
# ============================================================

# Left region
N20 = TS.Node(P20, DOF0)
N21 = TS.Node(P21, DOF0)
N22 = TS.Node(P22, DOF0)
N23 = TS.Node(P23, DOF0)

# Central region
N24 = TS.Node(P24, DOF0)
N25 = TS.Node(P25, DOF0)
N26 = TS.Node(P26, DOF0)
N27 = TS.Node(P27, DOF0)

N28 = TS.Node(P28, DOF0)
N29 = TS.Node(P29, DOF0)
N30 = TS.Node(P30, DOF0)
N31 = TS.Node(P31, DOF0)

# Right region
N32 = TS.Node(P32, DOF0)
N33 = TS.Node(P33, DOF0)
N34 = TS.Node(P34, DOF0)
N35 = TS.Node(P35, DOF0)


# ============================================================
# SUPPORTS
# ============================================================

# Left support
S1 = TS.Support(
    N1,
    TS.DOF(1, 1, 1, 0, 0, 0)
)

# Right support
S2 = TS.Support(
    N2,
    TS.DOF(0, 1, 0, 0, 0, 0)
)
```

```python
T2 = TS.TrussSystem()


# ============================================================
# SEARCH SPACE
# ============================================================

T2.addPolygontoSS(Polygon)

T2.addNode([
    # Supports
    N1, N2,

    # Top nodes
    N3, N4, N5, N6,
    N7, N8, N9, N10,
    N11, N12, N13, N14, N15,
    N16, N17, N18, N19,

    # Bottom / internal nodes
    N20, N21, N22, N23,
    N24, N25, N26, N27,
    N28, N29, N30, N31,
    N32, N33, N34, N35
])

T2.addMaterial(matC)  # struts
T2.addMaterial(matT)  # ties


# ============================================================
# EDGES
# ============================================================


# ============================================================
# 1. BOTTOM TIE - LEFT
# ============================================================

e1 = TS.Edge(N1,  N20, matT)
e2 = TS.Edge(N20, N21, matT)
e3 = TS.Edge(N21, N22, matT)
e4 = TS.Edge(N22, N23, matT)


# ============================================================
# 2. BOTTOM TIE - RIGHT
# ============================================================

e5 = TS.Edge(N32, N33, matT)
e6 = TS.Edge(N33, N34, matT)
e7 = TS.Edge(N34, N35, matT)
e8 = TS.Edge(N35, N2,  matT)


# ============================================================
# 3. BOTTOM TIE - CENTRAL
# ============================================================

e9 = TS.Edge(N23, N32, matT)


# ============================================================
# 4. TOP STRUT - LEFT
# ============================================================
#
# N3 ---- N4 ---- N5 ---- N6 ---- N7
#
# ============================================================

e10 = TS.Edge(N3, N4, matC)
e11 = TS.Edge(N4, N5, matC)
e12 = TS.Edge(N5, N6, matC)
e13 = TS.Edge(N6, N7, matC)


# ============================================================
# 5. TOP STRUT - CENTRAL
# ============================================================

e14 = TS.Edge(N7,  N8,  matC)
e15 = TS.Edge(N8,  N9,  matC)
e16 = TS.Edge(N9,  N10, matC)
e17 = TS.Edge(N10, N11, matC)
e18 = TS.Edge(N11, N12, matC)
e19 = TS.Edge(N12, N13, matC)
e20 = TS.Edge(N13, N14, matC)
e21 = TS.Edge(N14, N15, matC)


# ============================================================
# 6. TOP STRUT - RIGHT
# ============================================================

e22 = TS.Edge(N15, N16, matC)
e23 = TS.Edge(N16, N17, matC)
e24 = TS.Edge(N17, N18, matC)
e25 = TS.Edge(N18, N19, matC)


# ============================================================
# 7. VERTICALS - LEFT
# ============================================================

e26 = TS.Edge(N20, N3, matT)
e27 = TS.Edge(N21, N4, matT)
e28 = TS.Edge(N22, N5, matT)
e29 = TS.Edge(N23, N6, matT)


# ============================================================
# 8. DIAGONALS - LEFT
# ============================================================

e30 = TS.Edge(N1,  N3, matC)
e31 = TS.Edge(N20, N4, matC)
e32 = TS.Edge(N21, N5, matC)
e33 = TS.Edge(N22, N6, matC)
e34 = TS.Edge(N23, N7, matC)


# ============================================================
# 9. CENTRAL LOWER CHORD ABOVE OPENING
# ============================================================
#
# N24 -- N25 -- N26 -- N27 -- N28 -- N29 -- N30 -- N31
#
# There is no node at x = 3150 mm in this chord.
#
# ============================================================

e35 = TS.Edge(N24, N25, matT)
e36 = TS.Edge(N25, N26, matT)
e37 = TS.Edge(N26, N27, matT)

# crosses the position x = 3150, where no node exists
e38 = TS.Edge(N27, N28, matT)

e39 = TS.Edge(N28, N29, matT)
e40 = TS.Edge(N29, N30, matT)
e41 = TS.Edge(N30, N31, matT)


# ============================================================
# 10. CENTRAL VERTICALS
# ============================================================

e42 = TS.Edge(N24, N7,  matT)
e43 = TS.Edge(N25, N8,  matT)
e44 = TS.Edge(N26, N9,  matT)
e45 = TS.Edge(N27, N10, matT)

# No vertical at x = 3150 mm

e46 = TS.Edge(N28, N12, matT)
e47 = TS.Edge(N29, N13, matT)
e48 = TS.Edge(N30, N14, matT)
e49 = TS.Edge(N31, N15, matT)


# ============================================================
# 11. CENTRAL DIAGONALS
# ============================================================

# Left half
e50 = TS.Edge(N24, N8,  matC)
e51 = TS.Edge(N25, N9,  matC)
e52 = TS.Edge(N26, N10, matC)
e53 = TS.Edge(N27, N11, matC)

# Right half
e54 = TS.Edge(N28, N11, matC)
e55 = TS.Edge(N29, N12, matC)
e56 = TS.Edge(N30, N13, matC)
e57 = TS.Edge(N31, N14, matC)


# ============================================================
# 12. VERTICALS - RIGHT
# ============================================================

e58 = TS.Edge(N32, N16, matT)
e59 = TS.Edge(N33, N17, matT)
e60 = TS.Edge(N34, N18, matT)
e61 = TS.Edge(N35, N19, matT)


# ============================================================
# 13. DIAGONALS - RIGHT
# ============================================================

e62 = TS.Edge(N32, N15, matC)
e63 = TS.Edge(N33, N16, matC)
e64 = TS.Edge(N34, N17, matC)
e65 = TS.Edge(N35, N18, matC)

# Final diagonal to right support
e66 = TS.Edge(N19, N2, matC)


# ============================================================
# ADD EDGES
# ============================================================

T2.addEdge([

    # Bottom tie - left
    e1, e2, e3, e4,

    # Bottom tie - right
    e5, e6, e7, e8,

    # Bottom tie - central
    e9,

    # Top chord
    e10, e11, e12, e13,
    e14, e15, e16, e17,
    e18, e19, e20, e21,
    e22, e23, e24, e25,

    # Left verticals
    e26, e27, e28, e29,

    # Left diagonals
    e30, e31, e32, e33, e34,

    # Central lower chord
    e35, e36, e37, e38,
    e39, e40, e41,

    # Central verticals
    e42, e43, e44, e45,
    e46, e47, e48, e49,

    # Central diagonals
    e50, e51, e52, e53,
    e54, e55, e56, e57,

    # Right verticals
    e58, e59, e60, e61,

    # Right diagonals
    e62, e63, e64, e65, e66
])


# ============================================================
# EXTERNAL FORCES
#
# Distributed load q = 30 kN/m
# Units: N
# ============================================================

DOF_F = TS.DOF(0, 1, 0, 0, 0, 0)


# ============================================================
# LEFT REGION
# ============================================================

Force1 = TS.Force_ext(
    N3,
    [0, -22500, 0, 0, 0, 0],
    DOF_F
)

Force2 = TS.Force_ext(
    N4,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force3 = TS.Force_ext(
    N5,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force4 = TS.Force_ext(
    N6,
    [0, -14625, 0, 0, 0, 0],
    DOF_F
)


# ============================================================
# CENTRAL REGION
# ============================================================

Force5 = TS.Force_ext(
    N7,
    [0, -10125, 0, 0, 0, 0],
    DOF_F
)

Force6 = TS.Force_ext(
    N8,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force7 = TS.Force_ext(
    N9,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force8 = TS.Force_ext(
    N10,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force9 = TS.Force_ext(
    N11,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force10 = TS.Force_ext(
    N12,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force11 = TS.Force_ext(
    N13,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force12 = TS.Force_ext(
    N14,
    [0, -4500, 0, 0, 0, 0],
    DOF_F
)

Force13 = TS.Force_ext(
    N15,
    [0, -10125, 0, 0, 0, 0],
    DOF_F
)


# ============================================================
# RIGHT REGION
# ============================================================

Force14 = TS.Force_ext(
    N16,
    [0, -14625, 0, 0, 0, 0],
    DOF_F
)

Force15 = TS.Force_ext(
    N17,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force16 = TS.Force_ext(
    N18,
    [0, -13500, 0, 0, 0, 0],
    DOF_F
)

Force17 = TS.Force_ext(
    N19,
    [0, -22500, 0, 0, 0, 0],
    DOF_F
)


# ============================================================
# ADD FORCES
# ============================================================

T2.addForce(Force1)
T2.addForce(Force2)
T2.addForce(Force3)
T2.addForce(Force4)

T2.addForce(Force5)
T2.addForce(Force6)
T2.addForce(Force7)
T2.addForce(Force8)
T2.addForce(Force9)
T2.addForce(Force10)
T2.addForce(Force11)
T2.addForce(Force12)
T2.addForce(Force13)

T2.addForce(Force14)
T2.addForce(Force15)
T2.addForce(Force16)
T2.addForce(Force17)


# ============================================================
# CHECK TOTAL LOAD
# ============================================================

total_force = (
    22500
    + 13500
    + 13500
    + 14625
    + 10125
    + 7 * 4500
    + 10125
    + 14625
    + 13500
    + 13500
    + 22500
)

print(
    'Total applied vertical load = ',
    total_force,
    'N'
)


# ============================================================
# SUPPORTS
# ============================================================

T2.addSupport(S1)
T2.addSupport(S2)


# ============================================================
# SOLVE
# ============================================================

T2.solveTruss(True)

print(
    'Is strut-and-tie model in equilibrium? ',
    T2.checkEquilibrium()
)
```
