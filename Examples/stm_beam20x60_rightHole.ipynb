```python
# ============================================================
# UNITS: N, mm, MPa
# ============================================================


# ============================================================
# GEOMETRY
# ============================================================

b = 6300.0
h = 600.0

c = 100.0

# Main levels of the truss
y_bottom = 50.0
y_tie    = 350.0
y_top    = 550.0


# ============================================================
# MATERIALS
# ============================================================

# Concrete
matC = TS.Material(
    30000,     # Ec
    40,        # fck
    0.003,     # epsilon_c
    -1         # compression
)

# Steel
matT = TS.Material(
    205000,    # Es
    435,       # fsy
    0.08,      # epsilon_s
    1          # tension
)


# ============================================================
# OUTER GEOMETRY
# ============================================================

PG1 = TS.Point(x=0, y=0, z=0)
PG2 = TS.Point(x=b, y=0, z=0)
PG3 = TS.Point(x=b, y=h, z=0)
PG4 = TS.Point(x=0, y=h, z=0)


# ============================================================
# OPENING - VIGA 5
# ============================================================

PG5 = TS.Point(x=500,  y=150, z=0)
PG6 = TS.Point(x=1700, y=150, z=0)
PG7 = TS.Point(x=1700, y=300, z=0)
PG8 = TS.Point(x=500,  y=300, z=0)


# ============================================================
# OUTER POLYGON
# ============================================================

Polygon = TS.Polygon()

Polygon.addPoint(PG1)
Polygon.addPoint(PG4)
Polygon.addPoint(PG3)
Polygon.addPoint(PG2)


# ============================================================
# OPENING POLYGON
# ============================================================

Opening = TS.Polygon()

Opening.addPoint(PG5)
Opening.addPoint(PG8)
Opening.addPoint(PG7)
Opening.addPoint(PG6)


# ============================================================
# DEFAULT DOF
# ============================================================

DOF0 = TS.DOF(0, 0, 0, 0, 0, 0)


# ============================================================
# STRUCTURAL POINTS
# ============================================================
#
# P1  -> P10 : bottom
# P11 -> P20 : intermediate
# P21 -> P38 : top
#
# ============================================================


# ============================================================
# BOTTOM POINTS
# y = 50 mm
# ============================================================

P1  = TS.Point(x=150,  y=y_bottom, z=0)   # left support
P2  = TS.Point(x=1800, y=y_bottom, z=0)
P3  = TS.Point(x=2250, y=y_bottom, z=0)
P4  = TS.Point(x=2700, y=y_bottom, z=0)

P5  = TS.Point(x=3600, y=y_bottom, z=0)
P6  = TS.Point(x=4050, y=y_bottom, z=0)
P7  = TS.Point(x=4500, y=y_bottom, z=0)
P8  = TS.Point(x=4950, y=y_bottom, z=0)
P9  = TS.Point(x=5400, y=y_bottom, z=0)

P10 = TS.Point(x=6150, y=y_bottom, z=0)   # right support


# ============================================================
# INTERMEDIATE POINTS
# y = 350 mm
# ============================================================

P11 = TS.Point(x=600,  y=y_tie, z=0)
P12 = TS.Point(x=750,  y=y_tie, z=0)
P13 = TS.Point(x=900,  y=y_tie, z=0)
P14 = TS.Point(x=1050, y=y_tie, z=0)
P15 = TS.Point(x=1200, y=y_tie, z=0)
P16 = TS.Point(x=1350, y=y_tie, z=0)
P17 = TS.Point(x=1500, y=y_tie, z=0)
P18 = TS.Point(x=1650, y=y_tie, z=0)

P19 = TS.Point(x=1950, y=y_tie, z=0)
P20 = TS.Point(x=2070, y=y_tie, z=0)


# ============================================================
# TOP POINTS
# y = 550 mm
# ============================================================

P21 = TS.Point(x=600,  y=y_top, z=0)
P22 = TS.Point(x=750,  y=y_top, z=0)
P23 = TS.Point(x=900,  y=y_top, z=0)
P24 = TS.Point(x=1050, y=y_top, z=0)
P25 = TS.Point(x=1200, y=y_top, z=0)
P26 = TS.Point(x=1350, y=y_top, z=0)
P27 = TS.Point(x=1500, y=y_top, z=0)
P28 = TS.Point(x=1650, y=y_top, z=0)

P29 = TS.Point(x=1800, y=y_top, z=0)
P30 = TS.Point(x=1950, y=y_top, z=0)

P31 = TS.Point(x=2250, y=y_top, z=0)
P32 = TS.Point(x=2700, y=y_top, z=0)
P33 = TS.Point(x=3150, y=y_top, z=0)
P34 = TS.Point(x=3600, y=y_top, z=0)
P35 = TS.Point(x=4050, y=y_top, z=0)
P36 = TS.Point(x=4500, y=y_top, z=0)
P37 = TS.Point(x=4950, y=y_top, z=0)
P38 = TS.Point(x=5400, y=y_top, z=0)


# ============================================================
# CREATE NODES
# ============================================================


# ============================================================
# BOTTOM NODES
# ============================================================

N1  = TS.Node(P1,  DOF0)
N2  = TS.Node(P2,  DOF0)
N3  = TS.Node(P3,  DOF0)
N4  = TS.Node(P4,  DOF0)
N5  = TS.Node(P5,  DOF0)
N6  = TS.Node(P6,  DOF0)
N7  = TS.Node(P7,  DOF0)
N8  = TS.Node(P8,  DOF0)
N9  = TS.Node(P9,  DOF0)
N10 = TS.Node(P10, DOF0)


# ============================================================
# INTERMEDIATE NODES
# ============================================================

N11 = TS.Node(P11, DOF0)
N12 = TS.Node(P12, DOF0)
N13 = TS.Node(P13, DOF0)
N14 = TS.Node(P14, DOF0)
N15 = TS.Node(P15, DOF0)
N16 = TS.Node(P16, DOF0)
N17 = TS.Node(P17, DOF0)
N18 = TS.Node(P18, DOF0)

N19 = TS.Node(P19, DOF0)
N20 = TS.Node(P20, DOF0)


# ============================================================
# TOP NODES
# ============================================================

N21 = TS.Node(P21, DOF0)
N22 = TS.Node(P22, DOF0)
N23 = TS.Node(P23, DOF0)
N24 = TS.Node(P24, DOF0)
N25 = TS.Node(P25, DOF0)
N26 = TS.Node(P26, DOF0)
N27 = TS.Node(P27, DOF0)
N28 = TS.Node(P28, DOF0)

N29 = TS.Node(P29, DOF0)
N30 = TS.Node(P30, DOF0)

N31 = TS.Node(P31, DOF0)
N32 = TS.Node(P32, DOF0)
N33 = TS.Node(P33, DOF0)
N34 = TS.Node(P34, DOF0)
N35 = TS.Node(P35, DOF0)
N36 = TS.Node(P36, DOF0)
N37 = TS.Node(P37, DOF0)
N38 = TS.Node(P38, DOF0)


# ============================================================
# SUPPORTS
# ============================================================

# Left support
S1 = TS.Support(
    N1,
    TS.DOF(1, 1, 1, 0, 0, 0)
)

# Right support
S2 = TS.Support(
    N10,
    TS.DOF(0, 1, 0, 0, 0, 0)
)
```

```python
T2 = TS.TrussSystem()


# ============================================================
# SEARCH SPACE
# ============================================================

T2.addPolygontoSS(Polygon)

T2.addNode([
    # Bottom
    N1, N2, N3, N4, N5,
    N6, N7, N8, N9, N10,

    # Intermediate
    N11, N12, N13, N14, N15,
    N16, N17, N18, N19, N20,

    # Top
    N21, N22, N23, N24, N25,
    N26, N27, N28, N29, N30,
    N31, N32, N33, N34, N35,
    N36, N37, N38
])

T2.addMaterial(matC)
T2.addMaterial(matT)


# ============================================================
# 1. BOTTOM TIE
# ============================================================

e1 = TS.Edge(N1, N2, matT)
e2 = TS.Edge(N2, N3, matT)
e3 = TS.Edge(N3, N4, matT)
e4 = TS.Edge(N4, N5, matT)
e5 = TS.Edge(N5, N6, matT)
e6 = TS.Edge(N6, N7, matT)
e7 = TS.Edge(N7, N8, matT)
e8 = TS.Edge(N8, N9, matT)
e9 = TS.Edge(N9, N10, matT)


# ============================================================
# 2. TOP COMPRESSION CHORD
# ============================================================
#
# N21 ---- N22 ---- ... ---- N38
#
# ============================================================

e10 = TS.Edge(N21, N22, matC)
e11 = TS.Edge(N22, N23, matC)
e12 = TS.Edge(N23, N24, matC)
e13 = TS.Edge(N24, N25, matC)
e14 = TS.Edge(N25, N26, matC)
e15 = TS.Edge(N26, N27, matC)
e16 = TS.Edge(N27, N28, matC)
e17 = TS.Edge(N28, N29, matC)

e18 = TS.Edge(N29, N30, matC)
e19 = TS.Edge(N30, N31, matC)

e20 = TS.Edge(N31, N32, matC)
e21 = TS.Edge(N32, N33, matC)
e22 = TS.Edge(N33, N34, matC)
e23 = TS.Edge(N34, N35, matC)
e24 = TS.Edge(N35, N36, matC)
e25 = TS.Edge(N36, N37, matC)
e26 = TS.Edge(N37, N38, matC)


# ============================================================
# 3. HORIZONTAL TIE ABOVE OPENING
# ============================================================
#
# N11 ---- N12 ---- ... ---- N19
#
# ============================================================

e27 = TS.Edge(N11, N12, matT)
e28 = TS.Edge(N12, N13, matT)
e29 = TS.Edge(N13, N14, matT)
e30 = TS.Edge(N14, N15, matT)
e31 = TS.Edge(N15, N16, matT)
e32 = TS.Edge(N16, N17, matT)
e33 = TS.Edge(N17, N18, matT)
e34 = TS.Edge(N18, N19, matT)


# ============================================================
# 4. VERTICAL TIES ABOVE OPENING
# ============================================================
#
# N21   N22   N23   ...   N28
#  |     |     |           |
# N11   N12   N13   ...   N18
#
# ============================================================

e35 = TS.Edge(N21, N11, matT)
e36 = TS.Edge(N22, N12, matT)
e37 = TS.Edge(N23, N13, matT)
e38 = TS.Edge(N24, N14, matT)
e39 = TS.Edge(N25, N15, matT)
e40 = TS.Edge(N26, N16, matT)
e41 = TS.Edge(N27, N17, matT)
e42 = TS.Edge(N28, N18, matT)

# x = 1950 mm
e43 = TS.Edge(N30, N19, matT)


# ============================================================
# 5. DIAGONAL STRUTS - LEFT REGION
# ============================================================

e44 = TS.Edge(N1, N21, matC)

e45 = TS.Edge(N11, N22, matC)
e46 = TS.Edge(N12, N23, matC)
e47 = TS.Edge(N13, N24, matC)
e48 = TS.Edge(N14, N25, matC)
e49 = TS.Edge(N15, N26, matC)
e50 = TS.Edge(N16, N27, matC)
e51 = TS.Edge(N17, N28, matC)
e52 = TS.Edge(N18, N29, matC)

e53 = TS.Edge(N29, N19, matC)


# ============================================================
# 6. N29 -> N2
# ============================================================

e54 = TS.Edge(N29, N2, matT)


# ============================================================
# 7. MAIN TRANSITION
# ============================================================
#
# N2 -> N31
#
# ============================================================

e55 = TS.Edge(N2, N31, matC)


# ============================================================
# 8. RIGHT REGION - VERTICAL TIES
# ============================================================

# x = 2250
e56 = TS.Edge(N31, N3, matT)

# x = 2700
e57 = TS.Edge(N32, N4, matT)

# No bottom node at x = 3150

# x = 3600
e58 = TS.Edge(N34, N5, matT)

# x = 4050
e59 = TS.Edge(N35, N6, matT)

# x = 4500
e60 = TS.Edge(N36, N7, matT)

# x = 4950
e61 = TS.Edge(N37, N8, matT)

# x = 5400
e62 = TS.Edge(N38, N9, matT)


# ============================================================
# 9. RIGHT REGION - DIAGONAL STRUTS
# ============================================================

# Left side toward midspan
e63 = TS.Edge(N3, N32, matC)
e64 = TS.Edge(N4, N33, matC)

# Right side
e65 = TS.Edge(N33, N5, matC)
e66 = TS.Edge(N34, N6, matC)
e67 = TS.Edge(N35, N7, matC)
e68 = TS.Edge(N36, N8, matC)
e69 = TS.Edge(N37, N9, matC)

# Final diagonal to right support
e70 = TS.Edge(N38, N10, matC)


# ============================================================
# 10. ADDITIONAL BARS AROUND N20
# ============================================================
#
# New numbering:
#
# N30 -------- N31
#   \           /
#    \         /
#     N19 --- N20
#              \
#               \
#                N2
#
# N20 lies on line N2-N31
#
# ============================================================

e71 = TS.Edge(N30, N20, matC)
e72 = TS.Edge(N31, N20, matC)
e73 = TS.Edge(N20, N2,  matC)
e74 = TS.Edge(N19, N20, matT)


# ============================================================
# ADD EDGES
# ============================================================

T2.addEdge([

    # Bottom tie
    e1, e2, e3, e4, e5,
    e6, e7, e8, e9,

    # Top compression chord
    e10, e11, e12, e13, e14,
    e15, e16, e17, e18, e19,
    e20, e21, e22, e23, e24,
    e25, e26,

    # Horizontal tie above opening
    e27, e28, e29, e30,
    e31, e32, e33, e34,

    # Vertical ties
    e35, e36, e37, e38,
    e39, e40, e41, e42, e43,

    # Left diagonal struts
    e44, e45, e46, e47,
    e48, e49, e50, e51,
    e52, e53,

    # N29 -> N2
    e54,

    # Main transition
    e55,

    # Right vertical ties
    e56, e57, e58, e59,
    e60, e61, e62,

    # Right diagonal struts
    e63, e64, e65, e66,
    e67, e68, e69, e70,

    # Additional bars
    e71, e72, e73, e74
])


# ============================================================
# EXTERNAL FORCES
# ============================================================

q = 30.0  # N/mm

DOF_F = TS.DOF(0, 1, 0, 0, 0, 0)


# ============================================================
# LOADS ON TOP NODES
# ============================================================
#
# N21 =  600 mm
# N22 =  750 mm
# N23 =  900 mm
# N24 = 1050 mm
# N25 = 1200 mm
# N26 = 1350 mm
# N27 = 1500 mm
# N28 = 1650 mm
# N29 = 1800 mm
# N30 = 1950 mm
# N31 = 2250 mm
# N32 = 2700 mm
# N33 = 3150 mm
# N34 = 3600 mm
# N35 = 4050 mm
# N36 = 4500 mm
# N37 = 4950 mm
# N38 = 5400 mm
#
# ============================================================

top_nodes = [
    N21, N22, N23, N24,
    N25, N26, N27, N28,
    N29, N30, N31, N32,
    N33, N34, N35, N36,
    N37, N38
]

top_x = [
    600.0,
    750.0,
    900.0,
    1050.0,
    1200.0,
    1350.0,
    1500.0,
    1650.0,
    1800.0,
    1950.0,
    2250.0,
    2700.0,
    3150.0,
    3600.0,
    4050.0,
    4500.0,
    4950.0,
    5400.0
]


# ============================================================
# LIMITS OF DISTRIBUTED LOAD
# ============================================================

x_load_start = 150.0
x_load_end   = 6150.0


# ============================================================
# CREATE EQUIVALENT NODAL LOADS
# ============================================================

Forces = []

for i in range(len(top_nodes)):

    if i == 0:

        x_right = (
            top_x[i] + top_x[i + 1]
        ) / 2.0

        tributary = x_right - x_load_start

    elif i == len(top_nodes) - 1:

        x_left = (
            top_x[i - 1] + top_x[i]
        ) / 2.0

        tributary = x_load_end - x_left

    else:

        x_left = (
            top_x[i - 1] + top_x[i]
        ) / 2.0

        x_right = (
            top_x[i] + top_x[i + 1]
        ) / 2.0

        tributary = x_right - x_left

    F = q * tributary

    force = TS.Force_ext(
        top_nodes[i],
        [0, -F, 0, 0, 0, 0],
        DOF_F
    )

    Forces.append(force)


# ============================================================
# CHECK TOTAL APPLIED LOAD
# ============================================================

total_force = 0.0

for i in range(len(top_nodes)):

    if i == 0:

        tributary = (
            (top_x[0] + top_x[1]) / 2.0
            - x_load_start
        )

    elif i == len(top_nodes) - 1:

        tributary = (
            x_load_end
            - (top_x[-2] + top_x[-1]) / 2.0
        )

    else:

        tributary = (
            (top_x[i] + top_x[i + 1]) / 2.0
            - (top_x[i - 1] + top_x[i]) / 2.0
        )

    total_force += q * tributary


print(
    'Total applied vertical load = ',
    total_force,
    'N'
)


# ============================================================
# ADD FORCES
# ============================================================

for force in Forces:
    T2.addForce(force)


# ============================================================
# ADD SUPPORTS
# ============================================================

T2.addSupport(S1)
T2.addSupport(S2)


# ============================================================
# SOLVE
# ============================================================

T2.solveTruss(True)

print(
    'Is strut-and-tie model in equilibrium? ',
    T2.checkEquilibrium()
)
```
